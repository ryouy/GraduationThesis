```
from pathlib import Path
import subprocess,sys,json
REPO=Path('/content/GraduationThesis-inference-audit')
if not (REPO/'.git').exists():
    subprocess.run(['git','clone','--depth','1','https://github.com/ryouy/GraduationThesis.git',str(REPO)],check=True)
OUT=Path('/content/inference-audit-20261104')
OUT.mkdir(exist_ok=True)
p=subprocess.Popen([sys.executable,'-u','1104_inference_audit/run.py','--out',str(OUT)],cwd=REPO,stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True)
with (OUT/'run.log').open('w') as log:
    for line in p.stdout:
        log.write(line);log.flush();print(line,end='',flush=True)
assert p.wait()==0
```
