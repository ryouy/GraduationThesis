## 1. 全50組のProbe識別性能
第24層のROC-AUC。暫定基準は AUC ≥ 0.80 かつ BH補正後 q ≤ 0.05。35組がAUC基準、33組が両基準を通過。基準通過でも今回は未採用の23組を区別します。

````python
colors={'Selected':'#245a81','Passed, not selected':'#a6bdce','Failed AUC':'#b9b9b9','Failed FDR':'#b37c42'}
for i in range(2):
    d=pairs.iloc[i*25:(i+1)*25].copy().iloc[::-1]
    labels=d.ID+'  '+d.X+' / '+d.Y
    fig,ax=plt.subplots(figsize=(11,8.7),layout='constrained')
    ax.barh(labels,d['L24 AUC'],color=d.Status.map(colors),height=.72)
    ax.axvline(.8,color='#555555',ls='--',lw=1,label='AUC threshold')
    for j,v in enumerate(d['L24 AUC']): ax.text(v+.012,j,f'{v:.3f}',va='center',fontsize=9)
    ax.set(xlim=(0,1.12),xlabel='ROC-AUC (layer 24)',title=f'Concept-pair screening | {i*25+1}–{(i+1)*25} of 50')
    ax.set_xticks([0,.2,.4,.6,.8,1]); ax.grid(axis='x',alpha=.18)
    ax.legend(handles=[Patch(color=c,label=k) for k,c in colors.items()]+[Line2D([0],[0],color='#555555',ls='--',label='AUC threshold')],loc='lower center',bbox_to_anchor=(.5,-.15),ncol=3,frameon=False,fontsize=9)
    finish(fig,f'auc_pairs_{i+1}')
display(pairs)
````

## 2. 採用した10組

````python
d=pairs[pairs.Status.eq('Selected')].copy().iloc[::-1]
fig,ax=plt.subplots(figsize=(9,4.8),layout='constrained')
ax.barh(d.X+' / '+d.Y,d['L24 AUC'],color='#245a81')
for j,v in enumerate(d['L24 AUC']): ax.text(v+.012,j,f'{v:.3f}',va='center')
ax.set(xlim=(0,1.12),xlabel='ROC-AUC (layer 24)',title='Selected concept pairs (n=10)'); ax.set_xticks([0,.2,.4,.6,.8,1]); ax.grid(axis='x',alpha=.18)
finish(fig,'selected_10_auc')
display(d.iloc[::-1])
````

## 3. Cue候補68文と採択24文の散布図
X＝bicycle、Y＝submarine。横軸・縦軸はそれぞれ定義文へのコサイン距離であり、埋め込み空間の2次元投影ではありません。距離は小さいほど近いことを示します。

````python
fig,ax=plt.subplots(figsize=(8,6),layout='constrained')
for selected,label,color,marker in [(False,'Not selected (44)','#b9b9b9','o'),(True,'Selected (24)','#245a81','o')]:
    d=cues[cues.selected.eq(selected)]
    ax.scatter(d.distance_x_rounded,d.distance_y_rounded,s=53 if selected else 35,c=color,marker=marker,label=label,alpha=.9,edgecolors='white',linewidths=.5)
for cid in ['C01_003','C01_006','C01_015']:
    r=cues.set_index('cue_id').loc[cid]; ax.annotate(cid,(r.distance_x_rounded,r.distance_y_rounded),xytext=(6,8),textcoords='offset points',fontsize=9)
ax.set(xlabel='Cosine distance to X (bicycle)',ylabel='Cosine distance to Y (submarine)',title='Cue screening | bicycle vs. submarine'); ax.legend(frameon=False); ax.grid(alpha=.18)
finish(fig,'bicycle_cue_distances')
````
